# 02 — Classical ML Regression

Experiment with classical regressors on the preprocessed data before finalizing the training script (`src/train_ml.py`).

```python
import sys
sys.path.append('..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from src.preprocessing import run as run_preprocessing
from src.train_ml import load_processed, get_candidate_models, evaluate
```

## Load processed data

Run `src/preprocessing.py` first (or call it below) to generate `data/processed/`.

```python
# run_preprocessing('../data/raw/housing.csv', '../data/processed')  # uncomment on first run
X_train, X_test, y_train, y_test = load_processed('../data/processed')
X_train.shape, X_test.shape
```

## Try candidate models

```python
results = {}
models = get_candidate_models()

for name, model in models.items():
    model.fit(X_train, y_train)
    metrics = evaluate(model, X_test, y_test)
    results[name] = metrics
    print(f"{name}: RMSE={metrics['rmse']:.2f}  MAE={metrics['mae']:.2f}  R2={metrics['r2']:.3f}")
```

## Compare results

```python
results_df = pd.DataFrame(results).T.sort_values('rmse')
results_df
```

```python
results_df['rmse'].plot(kind='bar', figsize=(8, 4), title='Test RMSE by model')
plt.ylabel('RMSE')
plt.show()
```

## Feature importance (tree-based model)

```python
rf = models['random_forest']
importances = pd.Series(rf.feature_importances_).sort_values(ascending=False)
importances.head(15).plot(kind='barh', figsize=(8, 6), title='Top feature importances (encoded features)')
plt.gca().invert_yaxis()
plt.show()
```
