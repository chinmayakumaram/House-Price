# 01 — Exploratory Data Analysis

Inspect the raw housing dataset: shape, types, missing values, target distribution, and correlations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
%matplotlib inline
```

```python
df = pd.read_csv('../data/raw/housing.csv')
df.head()
```

## Shape, dtypes, missing values

```python
print(df.shape)
df.info()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

## Target distribution (`price`)

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df['price'], kde=True, ax=ax[0])
ax[0].set_title('Price distribution')
sns.boxplot(x=df['price'], ax=ax[1])
ax[1].set_title('Price boxplot')
plt.tight_layout()
plt.show()
```

## Correlations with target

```python
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()['price'].sort_values(ascending=False)
corr
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df.corr(), annot=False, cmap='coolwarm')
plt.title('Feature correlation heatmap')
plt.show()
```

## Categorical feature overview

```python
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
for c in cat_cols:
    print(c, '->', df[c].nunique(), 'unique values')
    display(df[c].value_counts().head())
```
